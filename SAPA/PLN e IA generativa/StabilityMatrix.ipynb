```python
import os
from transformers import BlipProcessor, BlipForConditionalGeneration
from PIL import Image
import torch

def load_model():
    """Carga el modelo y el procesador de BLIP."""
    processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-base")
    model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-base")
    return processor, model

def generate_caption(image_path, processor, model):
    """Genera una descripción para una imagen dada."""
    image = Image.open(image_path).convert("RGB")
    inputs = processor(image, return_tensors="pt")
    with torch.no_grad():
        out = model.generate(**inputs)
    return processor.decode(out[0], skip_special_tokens=True)

def process_images_in_directory(directory):
    """Genera descripciones para todas las imágenes en un directorio."""
    processor, model = load_model()
    image_extensions = {".jpg", ".jpeg", ".png", ".bmp", ".gif"}
    
    for filename in os.listdir(directory):
        if any(filename.lower().endswith(ext) for ext in image_extensions):
            image_path = os.path.join(directory, filename)
            caption = generate_caption(image_path, processor, model)
            print(f"Imagen: {filename}\nDescripción: {caption}\n")

if __name__ == "__main__":
    image_directory = "./images"  # Cambia esto por la ruta a tu directorio
    process_images_in_directory(image_directory)
```
